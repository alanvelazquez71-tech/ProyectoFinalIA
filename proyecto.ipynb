```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import layers, Model, optimizers
from tensorflow.keras.regularizers import l2

# ============================================
# 🔹 CONFIGURACIÓN DE COLUMNAS
# ============================================
user_features_cols = [
    'age', 'preferred_budget', 'preferred_avg_temp',
    'culture', 'adventure', 'nature', 'beaches',
    'nightlife', 'cuisine', 'wellness', 'urban', 'seclusion'
]

city_features_cols = [
    'temp_avg_norm', 'temp_max_norm', 'culture_norm', 'adventure_norm',
    'nature_norm', 'beaches_norm', 'nightlife_norm', 'cuisine_norm',
    'wellness_norm', 'urban_norm', 'seclusion_norm', 'budget_level_num_norm'
]

# ============================================
# 🔹 MATRICES DE ENTRADA
# ============================================
X_users = df_users_normalizado[user_features_cols].values
X_cities = df_normalizado[city_features_cols].values
user_ids = df_users_normalizado['user_id'].values
city_ids = df_normalizado['city_id'].values

num_users, num_user_features = X_users.shape
num_cities, num_city_features = X_cities.shape

print(f"📊 Users: {num_users}, Cities: {num_cities}")

# ============================================
# 🔹 MODELO TWO-TOWER MEJORADO
# ============================================
def build_improved_two_tower_model(embedding_dim=64):
    """
    Modelo Two-Tower mejorado con:
    - Dropout para regularización
    - BatchNormalization para estabilidad
    - Menos capas pero más anchas
    - Sin L2 normalization (permite variabilidad en scores)
    """
    user_input = layers.Input(shape=(num_user_features,), name="user_input")
    city_input = layers.Input(shape=(num_city_features,), name="city_input")

    # 🔹 Torre de Usuario
    user_tower = layers.Dense(256, activation='relu', kernel_regularizer=l2(0.001))(user_input)
    user_tower = layers.BatchNormalization()(user_tower)
    user_tower = layers.Dropout(0.3)(user_tower)
    
    user_tower = layers.Dense(128, activation='relu', kernel_regularizer=l2(0.001))(user_tower)
    user_tower = layers.BatchNormalization()(user_tower)
    user_tower = layers.Dropout(0.2)(user_tower)
    
    user_embedding = layers.Dense(embedding_dim, activation='linear', name="user_emb")(user_tower)

    # 🔹 Torre de Ciudad
    city_tower = layers.Dense(256, activation='relu', kernel_regularizer=l2(0.001))(city_input)
    city_tower = layers.BatchNormalization()(city_tower)
    city_tower = layers.Dropout(0.3)(city_tower)
    
    city_tower = layers.Dense(128, activation='relu', kernel_regularizer=l2(0.001))(city_tower)
    city_tower = layers.BatchNormalization()(city_tower)
    city_tower = layers.Dropout(0.2)(city_tower)
    
    city_embedding = layers.Dense(embedding_dim, activation='linear', name="city_emb")(city_tower)

    # 🔹 Similaridad (sin normalización L2)
    dot = layers.Dot(axes=1, normalize=False)([user_embedding, city_embedding])
    output = layers.Activation('sigmoid')(dot)

    model = Model(inputs=[user_input, city_input], outputs=output)
    
    # Usar Adam con learning rate adaptativo
    model.compile(
        optimizer=optimizers.Adam(learning_rate=0.0005),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )
    
    return model

model = build_improved_two_tower_model(embedding_dim=64)
model.summary()

# ============================================
# 🔹 GENERACIÓN DE PARES CON MÁS NEGATIVOS
# ============================================
negative_sample_size = 50  # Aumentado de 20 a 50
train_pairs = []
train_labels = []

print("🔄 Generando pares de entrenamiento...")

for idx, user in df_users_normalizado.iterrows():
    visited_cities = user['visited_city_encoded']
    if not isinstance(visited_cities, list):
        visited_cities = [visited_cities]

    for visited_city in visited_cities:
        if visited_city not in df_normalizado['city_id'].values:
            continue
        
        city_idx = df_normalizado.index[df_normalizado['city_id'] == visited_city][0]
        
        # Par positivo
        train_pairs.append([idx, city_idx])
        train_labels.append(1)

        # Pares negativos
        negative_indices = df_normalizado.index[~df_normalizado['city_id'].isin(visited_cities)]
        
        if len(negative_indices) < negative_sample_size:
            negative_sample = negative_indices
        else:
            negative_sample = np.random.choice(negative_indices, size=negative_sample_size, replace=False)
        
        for neg_idx in negative_sample:
            train_pairs.append([idx, neg_idx])
            train_labels.append(0)

train_pairs = np.array(train_pairs)
train_labels = np.array(train_labels)

X_user_train = X_users[train_pairs[:, 0]]
X_city_train = X_cities[train_pairs[:, 1]]
y_train = train_labels

print(f"✅ Total pares: {len(train_pairs)} | Positivos: {train_labels.sum()} | Negativos: {len(train_labels) - train_labels.sum()}")
print(f"📊 Ratio positivo/negativo: 1:{(len(train_labels) - train_labels.sum()) / train_labels.sum():.1f}")

# ============================================
# 🔹 ENTRENAMIENTO CON CALLBACKS
# ============================================
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

early_stop = EarlyStopping(
    monitor='loss',
    patience=5,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor='loss',
    factor=0.5,
    patience=3,
    min_lr=1e-6,
    verbose=1
)

print("\n🚀 Entrenando modelo...")
history = model.fit(
    [X_user_train, X_city_train], 
    y_train,
    epochs=50,
    batch_size=256,
    validation_split=0.1,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)

# ============================================
# 🔹 FUNCIÓN DE RECOMENDACIÓN MEJORADA
# ============================================
def recommend_top5_improved(user_id, top_n=5, show_all_scores=False):
    """
    Recomendaciones mejoradas con:
    - Scores raw del modelo (sin normalización extra)
    - Mejor visualización de diferencias
    - Opción de ver distribución completa de scores
    """
    if user_id not in df_users["user_id"].values:
        print(f"⚠️ El usuario {user_id} no existe en df_users.")
        return

    user = df_users[df_users["user_id"] == user_id].iloc[0]
    user_age = int(user["age"])
    user_budget = float(user["preferred_budget"])
    user_temp = float(user["preferred_avg_temp"])
    user_features = user[["culture", "adventure", "nature", "beaches",
                          "nightlife", "cuisine", "wellness", "urban", "seclusion"]].values.astype(float)

    user_idx = np.where(user_ids == user_id)[0][0]
    user_vec = X_users[user_idx].reshape(1, -1)
    user_vecs = np.repeat(user_vec, len(X_cities), axis=0)

    # Obtener scores RAW del modelo
    scores = model.predict([user_vecs, X_cities], verbose=0).flatten()
    
    # Mostrar estadísticas de scores
    print(f"\n📊 Estadísticas de Scores:")
    print(f"   Min: {scores.min():.4f} | Max: {scores.max():.4f} | Mean: {scores.mean():.4f} | Std: {scores.std():.4f}")
    
    if show_all_scores:
        print(f"\n🔍 Distribución de scores (percentiles):")
        for p in [10, 25, 50, 75, 90, 95, 99]:
            print(f"   P{p}: {np.percentile(scores, p):.4f}")

    sorted_idx = np.argsort(scores)[::-1]

    top_results = []
    seen_city_ids = set()

    for idx in sorted_idx:
        cid = city_ids[idx]
        if cid in seen_city_ids:
            continue

        city_row = df[df["id"] == cid].iloc[0]

        # Temperatura
        try:
            temp_data = city_row["avg_temp_monthly"]
            if isinstance(temp_data, str) and temp_data.startswith("{"):
                import json
                temp_dict = json.loads(temp_data)
                city_temp = np.mean([m["avg"] for m in temp_dict.values()])
            else:
                city_temp = float(temp_data)
        except Exception:
            city_temp = np.nan

        # Presupuesto
        budget_str = str(city_row["budget_level"]).lower()
        if "low" in budget_str or "budget" in budget_str:
            city_budget = 1.0
        elif "mid" in budget_str:
            city_budget = 2.0
        elif "high" in budget_str or "luxury" in budget_str:
            city_budget = 3.0
        else:
            try:
                city_budget = float(city_row["budget_level"])
            except:
                city_budget = np.nan

        # Filtro ±20%
        if not np.isnan(city_temp) and not np.isnan(user_temp):
            if abs(city_temp - user_temp) / max(abs(user_temp), 1e-6) > 0.2:
                continue
        
        if not np.isnan(city_budget) and not np.isnan(user_budget):
            if abs(city_budget - user_budget) / max(abs(user_budget), 1e-6) > 0.2:
                continue

        seen_city_ids.add(cid)
        top_results.append((idx, scores[idx]))
        
        if len(top_results) == top_n:
            break

    print(f"\n{'='*80}")
    print(f"🎯 TOP {top_n} RECOMENDACIONES PARA USUARIO {user_id}")
    print(f"{'='*80}")
    print(f"👤 Edad: {user_age} años | 💰 Presupuesto: {user_budget:.2f} | 🌡️ Temp. preferida: {user_temp:.2f}°C")
    print(f"{'='*80}\n")

    if not top_results:
        print("⚠️ No se encontraron ciudades que coincidan con las preferencias.")
        return

    for i, (idx, score) in enumerate(top_results):
        city_row = df[df["id"] == city_ids[idx]].iloc[0]

        # Temperatura recalculada
        try:
            temp_data = city_row["avg_temp_monthly"]
            if isinstance(temp_data, str) and temp_data.startswith("{"):
                import json
                temp_dict = json.loads(temp_data)
                city_temp = np.mean([m["avg"] for m in temp_dict.values()])
            else:
                city_temp = float(temp_data)
        except Exception:
            city_temp = np.nan

        # Top 3 features similares
        city_features = city_row[["culture", "adventure", "nature", "beaches",
                                  "nightlife", "cuisine", "wellness", "urban", "seclusion"]].values.astype(float)
        diffs = -np.abs(user_features - city_features)
        top3_idx = np.argsort(diffs)[-3:]
        feature_names = ["Cultura", "Aventura", "Naturaleza", "Playas", "Nightlife",
                         "Gastronomía", "Bienestar", "Urbano", "Aislamiento"]
        top3_features = [feature_names[j] for j in top3_idx[::-1]]

        # Barra de score visual
        bar_length = int(score * 50)
        score_bar = "█" * bar_length + "░" * (50 - bar_length)

        print(f"🏙️  #{i+1} | {city_row['city']} ({city_row['country']}, {city_row['region']})")
        print(f"   ⭐ Score: {score:.4f} [{score_bar}]")
        print(f"   🌡️  Temp: {city_temp:.1f}°C | 💰 {city_row['budget_level']}")
        print(f"   🎭 Cultura: {city_row['culture']:.1f} | 🏞️  Aventura: {city_row['adventure']:.1f}")
        print(f"   🌃 Nightlife: {city_row['nightlife']:.1f} | 🍽️  Gastronomía: {city_row['cuisine']:.1f}")
        print(f"   ✨ Coincidencias: {', '.join(top3_features)}")
        print(f"{'-'*80}\n")

# ============================================
# 🔹 EJECUTAR RECOMENDACIONES
# ============================================
recommend_top5_improved(user_id=1, top_n=5, show_all_scores=True)
```
